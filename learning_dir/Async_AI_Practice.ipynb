```python
from openai import AsyncOpenAI

async_openai_client = AsyncOpenAI() # this gives the kitchen the power to work concurrently. 

async def ask_llm(question:str) -> str:
    response = await async_openai_client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role":"system", "content":"You're concise."},
            {"role":"user", "content": question}
        ],
        temperature=0.4
    )
    return response.choices[0].message.content
```

```python
import time, asyncio

questions = [
    "If humanity successfully establishes a permanent colony on Mars, what legal and economic system should govern its resources?",
    "Why did octopuses evolve three hearts and blue blood, while most vertebrates manage with just one heart and red blood?",
    "How does the concept of zero revolutionize both abstract mathematics and modern computer architecture?",
    "What architectural principles allow ancient Roman concrete structures, like the Pantheon, to survive intact for two millennia without steel reinforcement?",
    "If an artificial intelligence creates a work of art entirely on its own, who—if anyone—should hold the moral and intellectual property rights to it?"
]

start = time.perf_counter()
for q in questions:
    await ask_llm(q) # here 5 waiters waiting for the response
end = time.perf_counter()
print(f"elapsed time in sequential LLM calls : {(end-start):.2f} seconds")

start = time.perf_counter()
response = await asyncio.gather(*(ask_llm(q) for q in questions)) # here only one waiter waiting for all 5 calls result.
end = time.perf_counter()
print(f"elasped time in concurrent LLM calls : {(end-start):.2f} seconds")
```
